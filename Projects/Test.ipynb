```python
from pytrends.request import TrendReq

pytrend = TrendReq()
pytrend.build_payload(kw_list=["AI tools", "ChatGPT"])
trends = pytrend.interest_over_time()
print(trends)
```

```python
from pytrends.request import TrendReq

pytrend = TrendReq()
pytrend.build_payload(["AI tools"])
data = pytrend.interest_over_time()

for date, value in zip(data.index, data["AI tools"]):
    print(f"{date.date()}: {int(value)}")
```
