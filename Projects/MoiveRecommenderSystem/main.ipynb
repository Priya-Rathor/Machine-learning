```python
# removing null values row 
movies.dropna(inplace = True)
```

```python
# getting array in genres
```

```python
import ast
def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
movies['genres']=movies['genres'].apply(convert)
```

```python
movies.head()
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies['cast'] = movies['cast'].apply(convert)
```

```python
movies.head()
```

```python
def convert3(cast_list):
    return (cast_list + [None]*3)[:3]
```

```python
movies['cast'] = movies['cast'].apply(convert3)
```

```python
movies.head()
```

```python

def fetch_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] =='Director':
          L.append(i['name'])
          break
    return L
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies.head()
```

```python
movies['overview'][0]
```

```python
movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
movies['genres']=movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies['Keywords']=movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies['cast'] = movies['cast'].apply(
    lambda x: [i.replace(" ", "") for i in x if isinstance(i, str)] if isinstance(x, list) else []
)

```

```python
movies['crew']=movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies.head()
```

```python

movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
new = movies.drop(columns=['overview','genres','keywords','cast','crew','Keywords'])
```

```python
new.head()
```

```python
new['tag'] = new['tags'].apply(lambda x:" ".join(x))
```

```python
new.head()
```

```python
new_df = new.drop(columns = ['tags'])
```

```python
new_df
```

```python
new_df['tag']=new_df['tag'].apply(lambda x:x.lower())
```

```python
import nltk

from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y =[]
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_df['tag'] =new_df['tag'].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features =5000, stop_words ='english')
```

```python
cv.fit_transform(new_df['tag']).toarray()
```

```python
cv.fit_transform(new_df['tag']).toarray().shape
```

```python
vectors = cv.fit_transform(new_df['tag']).toarray()
```

```python
vectors[0]
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
cosine_similarity(vectors).shape
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity[0]
```

```python
def recommned(movies):
    movies_index = new_df[new_df['title'] == movies].index[0]
    distances = similarity[movies_index]
    movies_list  = sorted(list(enumerate(distances)), reverse=True, key=lambda x:x[1])[1:6]
    for i in movies_list:
        print(new_df.iloc[i[0]].title)
       
    
```

```python
recommned('Avatar')
```

```python
recommned('Batman Begins')
```
