```python
import re

def get_cheatsheet_prompt(content: str) -> str:
    """
    Generates a detailed prompt for an AI agent to create a beginner-friendly, colorful PDF cheat sheet
    for the specified Python library or topic.
    
    Parameters:
        content (str): The name of the Python library or topic (e.g., 'Pandas', 'NumPy', etc.)
    
    Returns:
        str: A formatted prompt string for generating the cheat sheet.
    """
    prompt = f"""
You are an expert Python educator and technical writer. Your task is to create a colorful, beginner-friendly PDF cheat sheet for the Python library '{content}'.

The cheat sheet must be easy to understand for beginners, structured clearly, and visually engaging.

📘 Title: {content} Cheat Sheet (Beginner Friendly)

Include the following sections:

1. 📌 Introduction
   - What is {content}?
   - What does it do, explained in simple words?

2. 🎯 Why and When to Use
   - Real-life use cases and practical examples.
   - Common domains and problems it solves.

3. 🧠 Core Concepts & Key Components
   - Important modules, data structures, or concepts with short, essay-style explanations.

4. 🛠️ Important Functions and Methods (Grouped by Category)
   - Example categories: File Handling, Data Processing, Visualization.
   - For each function:
     - `function_name(args)` — One-line explanation.
     - ➤ Example: `example_code_here`

5. ✅ Pros and ❌ Cons
   - List 3–5 for each.

6. ⚠️ Common Mistakes to Avoid
   - Beginner pitfalls, wrong assumptions, common errors.

7. 💡 Tips for Learning and Using Effectively
   - Practice ideas, best documentation/tutorials, debugging tips.

8. 📚 Glossary
   - Define tricky terms in beginner-friendly language.

📄 Output Format:
- Structure the content with section headers, subheaders, and spacing.
- Use monospaced font for code.
- Make the layout colorful and ready for conversion to PDF.

Only generate content. Do not include anything unrelated to the cheat sheet.
"""
    return prompt.strip()
```

```python
from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.enums import TA_CENTER
from reportlab.lib import colors
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer

def create_colored_pdf(text_sections, output_filename="CheatSheet.pdf"):
    doc = SimpleDocTemplate(output_filename, pagesize=A4, rightMargin=40, leftMargin=40, topMargin=40, bottomMargin=40)
    story = []

    styles = getSampleStyleSheet()

    # Custom styles
    title_style = ParagraphStyle(
        'Title',
        parent=styles['Title'],
        fontSize=24,
        textColor=colors.HexColor('#4F46E5'),  # Indigo
        alignment=TA_CENTER,
        spaceAfter=20
    )

    header_style = ParagraphStyle(
        'Header',
        parent=styles['Heading2'],
        fontSize=16,
        textColor=colors.HexColor('#10B981'),  # Green
        spaceBefore=12,
        spaceAfter=6,
    )

    body_style = ParagraphStyle(
        'Body',
        parent=styles['BodyText'],
        fontSize=11,
        leading=18,
        textColor=colors.HexColor('#374151'),  # Gray
    )

    code_style = ParagraphStyle(
        'Code',
        fontName='Courier',
        fontSize=9,
        backColor=colors.whitesmoke,
        textColor=colors.black,
        leftIndent=10,
        rightIndent=10,
        spaceBefore=4,
        spaceAfter=4,
    )

    # Title
    story.append(Paragraph("🌟 Seaborn Cheat Sheet", title_style))
    story.append(Spacer(1, 12))

    for section in text_sections:
        story.append(Paragraph(section['header'], header_style))
        story.append(Paragraph(section['body'], body_style))
        if section.get("code"):
            story.append(Paragraph(section['code'], code_style))
        story.append(Spacer(1, 10))

    doc.build(story)
    print(f"✅ PDF generated successfully: {output_filename}")

# 🧠 Example Seaborn content
sections = [
    {
        "header": "1. sns.scatterplot()",
        "body": "Creates a scatter plot. Use `hue`, `style`, or `size` to add visual distinctions.",
        "code": """
<pre>
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset("tips")
sns.scatterplot(x="total_bill", y="tip", data=tips, hue="day", style="time")
plt.show()
</pre>
"""
    },
    {
        "header": "2. sns.histplot()",
        "body": "Creates a histogram of univariate or bivariate distributions. Use `kde=True` for smooth curves.",
        "code": """
<pre>
sns.histplot(data=tips, x="total_bill", bins=20, kde=True)
plt.show()
</pre>
"""
    },
    {
        "header": "3. sns.boxplot()",
        "body": "Displays the distribution of a variable across categories with quartiles and outliers.",
        "code": """
<pre>
sns.boxplot(x="day", y="total_bill", data=tips, hue="sex")
plt.show()
</pre>
"""
    },
    {
        "header": "4. sns.heatmap()",
        "body": "Visualizes matrices with colored cells. Useful for correlation plots and grids.",
        "code": """
<pre>
import numpy as np
matrix = np.random.rand(5, 5)
sns.heatmap(matrix, annot=True, cmap="coolwarm")
plt.show()
</pre>
"""
    },
    {
        "header": "5. sns.pairplot()",
        "body": "Plots pairwise relationships across an entire DataFrame with optional `hue` grouping.",
        "code": """
<pre>
iris = sns.load_dataset("iris")
sns.pairplot(iris, hue="species", palette="bright")
plt.show()
</pre>
"""
    }
]

# 🚀 Generate the PDF
create_colored_pdf(sections)
```
