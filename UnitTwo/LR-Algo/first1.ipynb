```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df =pd.read_csv('placement (2).csv')
df
```

```python
df.head()
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('CGPA')
plt.ylabel('Package (in lpa)')
```

```python
X = df.iloc[:,0:1]
y = df.iloc[:,-1]
```

```python
y
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train,y_train)
```

```python
X_test
```

```python
X_test.iloc[0]
y_test.iloc[10]
```

```python
lr.predict(X_test.iloc[10].values.reshape(1,1))  # we need to reshape it because sklearn expexts it in a 2D Format 
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.scatter(X_train, lr.predict(X_train),color = 'red')
plt.xlabel('CGPA')
plt.ylabel('Package (in lpa )')
```
