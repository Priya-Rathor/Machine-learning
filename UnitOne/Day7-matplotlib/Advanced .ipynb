```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # not always needed, but good to import

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')  # ✅ correct way to get a 3D subplot

# Example 3D scatter plot
ax.scatter3D(batter['runs'], batter['avg'], batter['strike_rate'])

# Labels
ax.set_xlabel('Runs')
ax.set_ylabel('Average')
ax.set_zlabel('Strike Rate')
plt.title('3D Scatter: Player Performance')

plt.show()
```

# 3D line Plot

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # safe to import for 3D plotting

# Sample data
x = [0, 1, 5]
y = [0, 10, 13]
z = [0, 13, 20]

# Create 3D plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')  # ✅ Correct 3D subplot creation

# Plot your 3D scatter
ax.scatter3D(x, y, z, marker='+')  # Or use batter['runs'], etc.

# Labeling
ax.set_xlabel('X-axis (e.g., Runs)')
ax.set_ylabel('Y-axis (e.g., Average)')
ax.set_zlabel('Z-axis (e.g., Strike Rate)')
ax.plot3D(x,y,z,color='red')
plt.title('3D Scatter Plot Example')

plt.show()
```

# 3D Surface Plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx , yy = np.meshgrid(x,y)

xx.shape
```

```python
z = xx**2 + yy**2
z.shape
```

```python

fig = plt.figure(figsize=(12,8))

ax = fig.add_subplot(111,projection = '3d')

ax.plot_surface(xx,yy,z,cmap='viridis')



ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')
ax.set_title('3D Surface Plot')

plt.show()
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))
ax= fig.add_subplot(111,projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z = np.log(xx) + np.log(yy)

fig = plt.figure(figsize=(12,8))
ax= fig.add_subplot(111,projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

# Contour Plots

```python
fig = plt.figure(figsize=(12,8))

ax = fig.add_subplot(111,projection = '3d')

ax.plot_surface(xx,yy,z,cmap='viridis')



ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')
ax.set_title('3D Surface Plot')

plt.show()
```

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

p = ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```
