# Relational Plot
* Scatter Plot
* line plot

* 1. To see the statistical relation between 2 or more variables.
* 2. Bivariate Analysis

```python
import seaborn as sns
import matplotlib as plt
import plotly.express as px

tips = sns.load_dataset('tips')
```

```python
# scatter plot
# it is a axws level function
sns.scatterplot(data =tips, x = 'total_bill',y='tip')

```

```python
# relplot -> figure level -> square shape
sns.relplot(data=tips, x= 'total_bill',y='tip',kind='scatter')
```

```python
sns.relplot(data=tips, x= 'total_bill',y='tip',kind='line')
```

```python
# hue Peramater :-  
sns.relplot(data=tips, x= 'total_bill',y='tip',kind='scatter',hue='sex')
```

```python
sns.relplot(data=tips, x= 'total_bill',y='tip',kind='scatter',hue='sex',style='time',size='size')
```

```python
# line plot

gap = px.data.gapminder()
gap
```

```python
temp_df = gap[gap['country'] == 'India']
temp_df
```

```python
# axes level function
sns.lineplot(data=temp_df,x='year',y='lifeExp')
```

```python
# using the relplot
sns.relplot(data=temp_df, x='year',y='lifeExp',kind='line')
```

```python
temp_df = gap[gap['country'].isin(['India','Pakistan','China'])]
temp_df
```

```python
sns.relplot(kind='line',data=temp_df,x='year',y='lifeExp',hue='country')
```

```python
temp_df = gap[gap['country'].isin(['India','Brazil','Germany'])]
temp_df
```

```python
sns.relplot(kind='line',data=temp_df,x='year',y='lifeExp',hue='country',style='continent',size='continent')
```

```python
# facet plot -> figure level function -> work with relplot
# it will not worl with scatterplot and lineplot
sns.relplot(data=tips, x='total_bill', y='tip',kind='scatter' , col='sex',row='day')
```

```python
sns.relplot(data=gap,x='lifeExp',y='gdpPercap',kind='scatter',col='year',col_wrap=2)
```

# Kdeplot

```python
# kdeplot
# Rather than using  discrete bins , a KDE plot smooths the observations with a Gaussian Kernel , producing a continuous density estimate 

tips
```

```python
sns.kdeplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips,x='total_bill' , kind='kde')
```

# Rugplot

```python
# Plot marginal distributions by drawing ticks along the  x and y axes.
# This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way 

sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
```

```python
# Bivariate histogram
# a bivariate histogram bins the data within rectangles that tile the plot
# and then shows the count of observations within each rectangle with the fill color
sns.histplot(data=tips, x='total_bill',y='tip')
```

```python
# Bivariate Kdeplot
# a bivariate KDE plot smoothes the (x,y) observations with a 2D gaussian
sns.kdeplot(data=tips, x='total_bill',y='tip')
```

# Boxplot :-

A boxplot is a standerdized way of displaying the distribution of data based on a fiev number summary ("minimum", first quartile[Q1],median, third quartile[Q3] and "maximum"). It can tell you about your outlier and what their values are. Boxplots can also tell you if your data is symmetrical , how tigtly your data is grouped and if and how your data is skewed

```python
# box PLot 
sns.boxplot(data=tips,x='day',y='total_bill')
```

```python
# Using catplot
sns.catplot(data=tips,x='sex',y='total_bill',kind='box')
```

```python
#hue
sns.boxplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
# single boxplot -> numerical col
sns.boxplot(data=tips,y='total_bill')
```

* When there are multiple observations in each category , it also uses bootstrapping to compute a confidence interval around the estimate, which is plotted using error bars

* When there are multiple observations in each category , it also uses bootstrapping to compute a confidence interval around the estimate , which is plotted using error bars 

```python
# countplot:-
# A special case for the bar plot is when you want to show the number of observations in each category rather than computing a statistic for a second variable . this is similar to a histogram over a categorical, rather then quantitative , variable
# this is the graph of count function in pandas
sns.countplot(data=tips,x='sex',hue='day')
```

In the simplest invocation, both function draw a scatterplot of two variables ,x and y, and then fir the regression model y~x and plot the resulting regression line and a 95% confidence interval for that regression.

# Plotting Pairwise Relationship(PairGrid Vs Pairplot) 

```python
iris
```

```python
sns.pairplot(iris, hue='species')
```

```python
sns.pairplot(iris)
```

```python
#PairGrid 
g =sns.PairGrid(data=iris)
g.map(sns.scatterplot)
```

```python
g =sns.PairGrid(data=iris, hue='species')
g.map(sns.scatterplot)
```

```python
# map_diag -> map_offdiag
g= sns.PairGrid(data=iris)
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
```

```python
g= sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
```

```python
g= sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.violinplot)
g.map_offdiag(sns.scatterplot)
```

```python
g= sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.boxenplot)
g.map_offdiag(sns.scatterplot)
```

```python
g= sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.boxplot)
g.map_offdiag(sns.scatterplot)
```

```python
g= sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.histplot)
g.map_offdiag(sns.histplot)
```

```python
g= sns.PairGrid(data=iris,hue='species')

g.map_diag(sns.histplot)
g.map_offdiag(sns.kdeplot)
```

```python
# map _diag -> map_upper -> map_lower
g= sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

```python
# vars
g= sns.PairGrid(data=iris,hue='species',vars=['sepal_width','petal_width'])
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```
