# Asking Question After see the data set 

## Understand the Context
 ` Before Touching the data ask`

* 1.What is the goal of this analysis ?
* 2. Who is the audience (Business, product team , researchers)?
* 3.What does each `row ` represent?
* 4. What do the `columns` mean?

# 2.Ask Descriptive Question (what happened?)

### these help you  summarize the dataset:

* 1. How many rows and columns are there ?
* 2. What there missing values?
* 3. What is the range of values in each column?
* 4. what are the top 5 frequent values in each category?

# 3. Ask Diagnostic Questions (why did it happen?)

## These help you find relationships or patterns:

* 1. Why are some  values missing?
* 2. Why did sales drop in a certain month?
* 3. What features might be influencing the outcome?
